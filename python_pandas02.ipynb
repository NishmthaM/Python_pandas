**Groupby in pandas**  # can do the group of data


```python
import pandas as pd
```

```python
a= pd.DataFrame({ "name": ["Bob","Steve","Will","Robin","Dustin","Mike"],
                  "Subject": [1,2,3,4,5,6],
                  "Subject_02":[10,20,30,40,50,60]})
a
```

```python
b=a.groupby("name")
b
```

```python
for x,y in b:
  print(x)
  print()
  print(y)
  print()
```

```python
# to get one group data

b.get_group("Bob")
```

```python
b.get_group("Mike")
```
