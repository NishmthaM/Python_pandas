**Groupby in pandas**  # can do the group of data


```python
import pandas as pd
```

```python
a= pd.DataFrame({ "name": ["Bob","Steve","Will","Robin","Dustin","Mike"],
                  "Subject": [1,2,3,4,5,6],
                  "Subject_02":[10,20,30,40,50,60]})
a
```

```python
b=a.groupby("name")
b
```

```python
for x,y in b:
  print(x)
  print()
  print(y)
  print()
```

```python
# to get one group data

b.get_group("Bob")
```

```python
b.get_group("Mike")
```

```python
a.max()
```

**Join and Append in pandas**

```python
# Join funciton

import pandas as pd

a= pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]})
b= pd.DataFrame({"C":[10,20,30,40,55],"D":[60,70,80,90,100]})

a.join(b)
```

```python
# Can also work in the unmatched data/ values

import pandas as pd

a= pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]})
b= pd.DataFrame({"C":[10,20,30],"D":[60,70,80]})

a.join(b)
```

```python

# can change the index value

import pandas as pd

a= pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]},index=["a","b","c","d","e"])
b= pd.DataFrame({"C":[10,20,30],"D":[60,70,80]},index=["a","b","c"])

b.join(a)
```

```python
a.join(b)
```

```python
b.join(a,how="right")
```

```python
b.join(a,how="left")
```

```python
b.join(a,how="outer")  # union
```

```python
b.join(a,how="inner")  # intersecton
```
